# Etiquetado / revisión de tiles

Requiere `uv sync --extra label` y abrir con `uv run jupyter lab`. Las cajas se autoguardan en cada edición.

```python
import glob, os
from maleza.common import load_config
from maleza.labeling import etiquetar

cfg = load_config("../config.yaml")
tiles_dir, labels_dir = cfg["paths"]["tiles_dir"], cfg["paths"]["labels_dir"]
```

```python
# Etiquetar tiles nuevos (ej. del ortomosaico 2): arranca por el primero sin label
tiles = glob.glob(os.path.join(tiles_dir, "tile_o2_*.jpg"))
etiquetar(tiles, labels_dir)
```

```python
# Revisar tiles ya etiquetados (carga las cajas existentes, recuerda dónde quedaste)
tiles = [t for t in glob.glob(os.path.join(tiles_dir, "tile_[0-9]*.jpg"))]
etiquetar(tiles, labels_dir, solo_pendientes=False,
          progress_path=os.path.join(cfg["root"], "data", "revision_progress.json"))
```
